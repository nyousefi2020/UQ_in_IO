```python
dim = 20
sigma_actual = 0.1  # variance (y ~ N(x, sigma_actual*I)); must match run_sim.py
n_points = 1000  # must match run_sim.py  
n_constraints = 75  
n_simulations = 100
r_seed_base = 42

c_actual = np.ones(dim)

# "gurobi" | "highs" — must match run_sim (set ALG1_LP_BACKEND or lp_backend default there).
lp_backend = "gurobi"

data_dir = f"data/dim{dim}/"
os.makedirs(data_dir, exist_ok=True)
filename = os.path.join(data_dir, f"s{sigma_actual}_n{n_points}_m{n_constraints}_data.pkl")

data_dict = {}
for sim in range(n_simulations):
    np.random.seed(r_seed_base + sim)
    data = utils.generate_lp_data_var(
        dim, n_constraints, n_points, c_actual, sigma_actual, lp_backend=lp_backend
    )
    data_dict[sim] = data

with open(filename, "wb") as f:
    pickle.dump(data_dict, f)

print(f"Saved {filename}")
```
