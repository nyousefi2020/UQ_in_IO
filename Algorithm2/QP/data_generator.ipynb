```python
import numpy as np
import pandas as pd
import gurobipy as gp
from gurobipy import GRB
import math
from scipy.stats import multivariate_normal,halfcauchy,vonmises_fisher,gamma
import utils 
import pickle
import os
```

```python
dim = 20 
r = np.ones(dim)
k = -1
c_actual = utils.normalize_c(np.ones(dim))
kappa_actual = 5
n_points = 200
n_simulations = 100
```

```python
data_dict = {}
for sim in range(n_simulations):
    data = utils.generate_qp_data_var(dim, n_points, c_actual, kappa_actual, r, k)# (Q,y,c)
    data_dict[(sim)] = data

data_dir = os.path.join("data", f"dim{dim}")
if not os.path.exists(data_dir):
    os.makedirs(data_dir)
    
with open(os.path.join(data_dir, f'k{kappa_actual}_n{n_points}_data.pkl'), 'wb') as file:
    pickle.dump(data_dict, file)
```
