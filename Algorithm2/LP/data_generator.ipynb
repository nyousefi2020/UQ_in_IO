```python
import numpy as np
import pandas as pd
import gurobipy as gp
from gurobipy import GRB
import math
from scipy.stats import multivariate_normal,halfcauchy,vonmises_fisher,gamma
import utils 
import pickle
import os
```

```python
dim = 20
n_constraints = 75  #(>n+1 for feasibility)
c_actual = utils.normalize_c(np.ones(dim))
kappa_actual = 10
n_points = 1000
n_simulations = 100
```

```python
data_dict = {}
for sim in range(n_simulations):
    data = utils.generate_lp_data_var(dim, n_constraints, n_points, c_actual, kappa_actual) # (A,b,y,c)
    data_dict[(sim)] = data

data_dir = os.path.join("data", f"dim{dim}")
os.makedirs(data_dir, exist_ok=True)
filename = os.path.join(data_dir, f"k{kappa_actual}_n{n_points}_m{n_constraints}_data.pkl")

with open(filename, 'wb') as file:
    pickle.dump(data_dict, file)
```
