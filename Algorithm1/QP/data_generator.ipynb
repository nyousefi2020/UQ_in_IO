```python
import numpy as np
import pandas as pd
import gurobipy as gp
from gurobipy import GRB
import math
from scipy.stats import multivariate_normal,halfcauchy
import utils 
import pickle
import os
```

```python
dim = 20
sigma_actual = 0.1  # variance (y ~ N(x, sigma_actual*I))
n_points = 1000  # must match run_sim.py
n_simulations = 100
r_seed_base = 42
k = 0  

# Create r and c_actual
r = np.ones(dim)
c_actual = np.ones(dim)

# Folder structure
data_dir = f"data/dim{dim}/"
os.makedirs(data_dir, exist_ok=True)
filename = os.path.join(data_dir, f"s{sigma_actual}_n{n_points}_data.pkl")

# Generate data
data_dict = {}
for sim in range(n_simulations):
    np.random.seed(r_seed_base + sim)
    data = utils.generate_qp_data_var(dim, n_points, c_actual, sigma_actual, r, k)
    data_dict[sim] = data

# Save
with open(filename, 'wb') as file:
    pickle.dump(data_dict, file)
```
